```python
email_address = "student.name_group@lpnu.com"

def analyze_email(email_address):
    at_index = email_address.find("@")

    if at_index == -1:
        return None

    login = email_address[:at_index]
    domain = email_address[at_index + 1:]
    result = f"{email_address}, {login}, {domain}"
    print(result, end="\n")
    print(email_address, login, domain, sep=" | ")



analyze_email(email_address)
```

```python
def format_name(first_name, last_name, upper_case=False):
    full_name = f"{first_name} {last_name}"

    if upper_case ==True:
        full_name = full_name.upper()
    else: 
        full_name = full_name.title()
    return full_name


print(format_name("Nick", "Smith"))
print(format_name("Lia", "Johnson", upper_case = True))
```

```python
def create_shopping_list(store_name, *items):
    list_str= ""
    count = 1
    for item in items:
        list_str += f"{count} {item}\n"
        count += 1
    return (store_name, list_str)    


print(*create_shopping_list("MegaMarket", "milk", "bread", "cheese", "apples"), 
      sep="\n---\n")
```

```python
def process_tags(tags_string, delimiter= ","):
    tags = tags_string.split(delimiter)

    chars_to_remove = " !--"
    tags = [tag.strip(chars_to_remove) for tag in tags]

    separator = "  #  "
    tags_string = separator.join(tags)

    return tags_string

tags_string = " python, data science! , --machine learning "
process_tags(tags_string)
```

```python
def create_profile(login, password, separator = "| ", **user_data):
    profile = f"login{separator}{login}\npassword{separator}{password}\n"

    for key, value in user_data.items():
        profile += f"{key}{separator}{value}\n"
    return profile


my_profile = create_profile("Student", "PI-21", separator = ": ", name='Ivan',
age=30, occupation='Developer', city='Lviv')
print(my_profile)
```
